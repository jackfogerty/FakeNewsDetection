```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import make_pipeline
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
# Set the path to the file you'd like to load
fake_path = "Data/Fake.csv"
real_path = "Data/True.csv"
tweets_df_train_path = "Data/train.csv"
tweets_df_test_path = "Data/test.csv"

# Load datasets
fake_df = pd.read_csv(fake_path)
fake_df["label"] = "fake"

real_df = pd.read_csv(real_path)
real_df["label"] = "real"

# Combine title and text
real_df["all_text"] = real_df["title"] + " " + real_df["text"]
fake_df["all_text"] = fake_df["title"] + " " + fake_df["text"]

# Load tweet dataset
tweets_df_train = pd.read_csv(tweets_df_train_path)
tweets_df_train = tweets_df_train.rename(columns={"target": "label"})
tweets_df_train = tweets_df_train.rename(columns={"text": "all_text"})
tweets_df_train["label"] = tweets_df_train["label"].map({0: "real", 1: "fake"})

# Combine all datasets
df = pd.concat([fake_df, real_df, tweets_df_train], ignore_index=True)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    df["all_text"], df["label"], test_size=0.2, random_state=42
)
```

```python
# Build Linear SVM pipeline
model = make_pipeline(
    TfidfVectorizer(),
    LinearSVC()
)

# Train model
model.fit(X_train, y_train)

# Evaluate
accuracy = model.score(X_test, y_test)
print(f"Linear SVM Accuracy: {accuracy:.2%}")

# More detailed metrics
y_pred = model.predict(X_test)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
