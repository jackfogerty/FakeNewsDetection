```python
#  ------------- IMPORT REQUIRED LIBRARIES -------------

# TfidfVectorizer will convert the text into numerical features that are based on word importance.
from sklearn.feature_extraction.text import TfidfVectorizer
# LinearSVC is a Support Vector Machine classifier that will help with text classification.
from sklearn.svm import LinearSVC
# make_pipeline will help us put together preprocessing and model training into 1 workflow. 
from sklearn.pipeline import make_pipeline
# pandas will be used to load and manipulate datasets.
import pandas as pd
# train_test_split will help us split the data into training and testing parts. 
from sklearn.model_selection import train_test_split
# classification_report will give us precision, recall, and F1-score for each class.
from sklearn.metrics import classification_report
```

```python
big_path = "Data/big_data.csv"
smaller_path = "Data/smaller_data.csv"

df = pd.read_csv(smaller_path)

df['label'] = df['label'].map({0: 'real', 1: 'fake'})

X_train, X_test, y_train, y_test = train_test_split(
    df['text'], df['label'], test_size=0.2, random_state=42
)
```

```python
#  ------------- BUILD THE Linear SVM PIPELINE -------------
# 1: TfidfVectorizer will make the text into numerical vectory by using TF-IDF. 
# 2: LinearSVC will make those vectors so it can classify text as real or fake. 
model = make_pipeline(
    TfidfVectorizer(),
    LinearSVC()
)

#  ------------- TRAIN THE MODEL -------------
# Here the model will learn patterns from the training data. 
model.fit(X_train, y_train)

#  ------------- EVALUATE MODEL ACCURACY -------------
# score() will return how accurate the test set is.
accuracy = model.score(X_test, y_test)
print(f"Linear SVM Accuracy: {accuracy:.2%}")

#  ------------- DETAILED METRICS -------------
# This will predict the labeld for the test set
y_pred = model.predict(X_test)
# This will show precision, recall, and F1-score for each class.
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
def predict_article(text):
    return model.predict([text])[0]

real_text = "Popular streamer Clavicular framemogged by ASU frat leader."
prediction = predict_article(real_text)

print(f"The model predicts the real text as: {prediction}")

fake_text = "Scientists discover that the moon is made out of cheese."
prediction = predict_article(fake_text)

print(f"The model predicts the fake text as: {prediction}")
```
